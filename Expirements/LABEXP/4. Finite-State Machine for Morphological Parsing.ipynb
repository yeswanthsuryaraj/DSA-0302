```python
def pluralize(word):
    state = 0

    if word.endswith(("s", "x", "z", "ch", "sh")):
        state = 1
        return word + "es"

    elif word.endswith("y") and len(word) > 1:
        if word[-2].lower() not in "aeiou":
            state = 2
            return word[:-1] + "ies"

    else:
        state = 3
        return word + "s"


nouns = ["cat", "dog", "bus", "box", "church", "brush", "baby", "city"]

for noun in nouns:
    plural = pluralize(noun)
    print(noun, "->", plural)
```

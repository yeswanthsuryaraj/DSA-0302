```python
import random

text = "i like tea i like coffee i like tea"
words = text.split()

bigrams = {}

for i in range(len(words) - 1):
    word = words[i]
    next_word = words[i + 1]
    bigrams.setdefault(word, []).append(next_word)

word = "i"
result = [word]

for i in range(5):
    if word not in bigrams:
        break
    word = random.choice(bigrams[word])
    result.append(word)

print("Generated text:", " ".join(result))
```

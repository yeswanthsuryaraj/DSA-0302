```python
def finite_automaton(string):
    state = 'q0'

    for char in string:
        if state == 'q0':
            if char == 'a':
                state = 'q1'
            else:
                state = 'q0'

        elif state == 'q1':
            if char == 'a':
                state = 'q1'
            elif char == 'b':
                state = 'q2'
            else:
                state = 'q0'

        elif state == 'q2':
            if char == 'a':
                state = 'q1'
            else:
                state = 'q0'

    return state == 'q2'

strings = ["ab", "cab", "aab", "abc", "helloab", "abab", "aba"]

for string in strings:
    if finite_automaton(string):
        print(string, "-> Accepted")
    else:
        print(string, "-> Rejected")
```

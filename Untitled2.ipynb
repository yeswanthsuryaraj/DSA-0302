```python
data = """5 3
b a !
0
1 4
5
0 b 1
1 a 2
2 a 3
3 a 3
3 ! 4
5
baa!
baaa!
ba!
b
baa"""

lines = iter(data.splitlines())

n, k = map(int, next(lines).split())
alphabet = next(lines).split()
start = int(next(lines))

parts = list(map(int, next(lines).split()))
accepting = set(parts[1:])

t = int(next(lines))
transitions = {}

for _ in range(t):
    p, c, r = next(lines).split()
    transitions.setdefault((int(p), c), set()).add(int(r))

Q = int(next(lines))

for _ in range(Q):
    query = next(lines).strip()
    current = {start}

    if query == "#":
        query = ""

    for ch in query:
        nxt = set()
        for state in current:
            nxt.update(transitions.get((state, ch), set()))
        current = nxt

    print("ACCEPT" if current & accepting else "REJECT")
```

```python
def inflect(stem, feature):
    vowels = "aeiou"

    def doublable(w):
        if len(w) < 3 or w[-1] in "wxy":
            return False
        if not (w[-3] not in vowels and w[-2] in vowels and w[-1] not in vowels):
            return False
        groups = 0
        inside = False
        for c in w:
            if c in vowels:
                if not inside:
                    groups += 1
                    inside = True
            else:
                inside = False
        return groups == 1

    if feature in ("N+PL", "V+3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and stem[-2] not in vowels:
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
            return stem[:-1] + "ies"
        return stem + "s"

    if feature == "V+PROG":
        if len(stem) >= 3 and stem.endswith("e") and stem[-2] not in vowels:
            return stem[:-1] + "ing"
        if doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    suffix = {
        "V+PAST": "ed",
        "ADJ+COMP": "er",
        "ADJ+SUP": "est"
    }.get(feature, "")

    if stem.endswith("e"):
        return stem + suffix[1:]
    if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
        return stem[:-1] + "i" + suffix
    if doublable(stem):
        return stem + stem[-1] + suffix
    return stem + suffix
```

```python
def is_doublable(stem):
    vowels = "aeiou"

    if len(stem) < 3 or stem[-1] in "wxy":
        return False

    if not (
        stem[-3] not in vowels
        and stem[-2] in vowels
        and stem[-1] not in vowels
    ):
        return False

    groups = 0
    inside = False

    for ch in stem:
        if ch in vowels:
            if not inside:
                groups += 1
                inside = True
        else:
            inside = False

    return groups == 1


def inflect(stem, feature):
    vowels = "aeiou"

    if feature in ("PL", "3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and stem[-2] not in vowels:
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
            return stem[:-1] + "ies"
        return stem + "s"

    if feature == "PROG":
        if len(stem) >= 3 and stem.endswith("e") and stem[-2] not in vowels:
            return stem[:-1] + "ing"
        if is_doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    suffix = {"PAST": "ed", "COMP": "er", "SUP": "est"}[feature]

    if stem.endswith("e"):
        return stem + suffix[1:]
    if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
        return stem[:-1] + "i" + suffix
    if is_doublable(stem):
        return stem + stem[-1] + suffix
    return stem + suffix
```

```python
stems = [
    ("cat", "N"),
    ("watch", "N"),
    ("watch", "V"),
    ("study", "V"),
    ("big", "ADJ")
]

features = {
    "N": ["PL"],
    "V": ["3SG", "PROG", "PAST"],
    "ADJ": ["COMP", "SUP"]
}

analyses = {}

for stem, pos in stems:
    analyses.setdefault(stem, []).append(stem + "+" + pos)

    for feature in features[pos]:
        surface = inflect(stem, feature)
        analysis = stem + "+" + pos + "+" + feature
        analyses.setdefault(surface, []).append(analysis)

queries = [
    "cats", "watches", "watching", "studied",
    "bigger", "cat", "biggest", "elephants"
]

for query in queries:
    results = sorted(analyses.get(query, []))
    print(", ".join(results) if results else "UNKNOWN")
```

```python
def is_consonant(word, i):
    ch = word[i]

    if ch in "aeiou":
        return False

    if ch == "y":
        if i == 0:
            return True
        return not is_consonant(word, i - 1)

    return True


def measure(word):
    m = 0
    i = 0
    n = len(word)

    while i < n and is_consonant(word, i):
        i += 1

    while i < n:
        while i < n and not is_consonant(word, i):
            i += 1

        if i == n:
            break

        while i < n and is_consonant(word, i):
            i += 1

        m += 1

    return m


def contains_vowel(word):
    return any(not is_consonant(word, i) for i in range(len(word)))


def double_consonant(word):
    return (
        len(word) >= 2
        and word[-1] == word[-2]
        and is_consonant(word, len(word) - 1)
    )


def cvc(word):
    if len(word) < 3:
        return False

    n = len(word)

    return (
        is_consonant(word, n - 3)
        and not is_consonant(word, n - 2)
        and is_consonant(word, n - 1)
        and word[-1] not in "wxy"
    )
```

```python
for word in ["tree", "trouble", "private", "toy", "syzygy"]:
    print(word, measure(word))
```
